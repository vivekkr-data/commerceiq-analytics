```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
DASH = ROOT / 'data' / 'processed' / 'dashboard'
metrics = pd.read_csv(DASH / 'model_metrics.csv')
metrics
```

## Delivery risk
The test period is the latest chronological 20% of delivered orders. Accuracy is not used alone because late deliveries are uncommon.

```python
delivery = metrics[metrics['module'].eq('Delivery Risk')]
delivery[['model', 'precision', 'recall', 'f1', 'roc_auc', 'pr_auc', 'selected']]
```

```python
roc = pd.read_parquet(DASH / 'delivery_roc_curve.parquet')
plt.figure(figsize=(5, 5))
plt.plot(roc['false_positive_rate'], roc['true_positive_rate'], label='Selected model')
plt.plot([0, 1], [0, 1], '--', label='Random')
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate'); plt.legend();
```

```python
forecast = metrics[metrics['module'].eq('Forecasting')]
forecast[['model', 'mae', 'rmse', 'mape', 'selected']].sort_values('rmse')
```
