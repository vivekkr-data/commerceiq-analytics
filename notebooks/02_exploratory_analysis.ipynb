# 02 — Exploratory Analysis
Business-focused exploration of delivered sales, categories, logistics, and reviews.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style='whitegrid')
ROOT = Path.cwd()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
DASH = ROOT / 'data' / 'processed' / 'dashboard'
orders = pd.read_parquet(DASH / 'overview.parquet')
delivered = orders[orders['order_status'].eq('delivered')].copy()
```

```python
monthly = delivered.groupby('purchase_period', as_index=False).agg(
    revenue=('merchandise_revenue', 'sum'), orders=('order_id', 'nunique'))
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
sns.lineplot(data=monthly, x='purchase_period', y='revenue', marker='o', ax=axes[0])
sns.lineplot(data=monthly, x='purchase_period', y='orders', marker='o', ax=axes[1])
axes[0].set_title('Monthly Delivered Merchandise Revenue')
axes[1].set_title('Monthly Delivered Orders')
plt.tight_layout();
```

```python
category = delivered.groupby('product_category')['merchandise_revenue'].sum().nlargest(12).sort_values()
category.plot.barh(figsize=(9, 5), title='Top Categories by Delivered Merchandise Revenue')
plt.xlabel('Revenue (R$)'); plt.tight_layout();
```

```python
delivered.groupby('late_delivery')['average_review_score'].agg(['count', 'mean']).rename(index={0: 'On time or early', 1: 'Late'})
```

The delivery/review comparison is observational. It supports an association, not a causal claim.
